```python
from pathlib import Path
from docling.document_converter import DocumentConverter
from tqdm import tqdm
```

```python
BASE_DIR = Path(".")

INPUT_DIR = BASE_DIR / "knowledge_base"
OUTPUT_DIR = BASE_DIR / "parsed_docs"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Knowledge Base:", INPUT_DIR)
print("Parsed Docs:", OUTPUT_DIR)
```

```python
converter = DocumentConverter()
```

```python
pdf_files = list(INPUT_DIR.rglob("*.pdf"))

print(f"Found {len(pdf_files)} PDFs\n")

for pdf in pdf_files:
    print(pdf)
```

```python
import fitz  # PyMuPDF
import time
import os

converter = DocumentConverter()

success = 0
failed = 0

print("=" * 80)
print(f"Found {len(pdf_files)} PDF(s) to process")
print("=" * 80)

for i, pdf_path in enumerate(pdf_files, start=1):

    print("\n" + "=" * 80)
    print(f"📄 PDF [{i}/{len(pdf_files)}]")
    print(f"Name      : {pdf_path.name}")

    try:
        pdf = fitz.open(pdf_path)
        total_pages = len(pdf)
        pdf.close()
    except:
        total_pages = "Unknown"

    file_size = os.path.getsize(pdf_path) / (1024 * 1024)

    print(f"Pages     : {total_pages}")
    print(f"Size      : {file_size:.2f} MB")
    print("=" * 80)

    start = time.time()

    try:

        print("⏳ Parsing document...")

        result = converter.convert(str(pdf_path))

        print("📝 Exporting markdown...")

        markdown = result.document.export_to_markdown()

        relative_path = pdf_path.relative_to(INPUT_DIR)

        output_file = OUTPUT_DIR / relative_path.with_suffix(".md")

        output_file.parent.mkdir(parents=True, exist_ok=True)

        output_file.write_text(markdown, encoding="utf-8")

        elapsed = time.time() - start

        success += 1

        print("\n✅ SUCCESS")
        print(f"Output File     : {output_file}")
        print(f"Characters      : {len(markdown):,}")
        print(f"Lines           : {markdown.count(chr(10)):,}")
        print(f"Processing Time : {elapsed:.2f} sec")

    except Exception as e:

        failed += 1

        print("\n❌ FAILED")
        print(e)

print("\n" + "=" * 80)
print("PARSING SUMMARY")
print("=" * 80)
print(f"✅ Success : {success}")
print(f"❌ Failed  : {failed}")
print(f"📚 Total   : {len(pdf_files)}")
print("=" * 80)
```
